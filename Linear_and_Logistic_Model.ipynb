```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import GroupShuffleSplit
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error

# Load data and define target
df = pd.read_csv('/content/drive/MyDrive/School work/TB_Burden_Country.csv') # Corrected file path
target_col = 'Estimated incidence (all forms) per 100 000 population'

# Drop rows where target is missing and drop entirely empty columns
df = df.dropna(subset=[target_col]).dropna(axis=1, how='all')
y = df[target_col]

# Feature Selection( clean x)
# Define leakage keywords and metadata to remove
forbidden_words = ['incidence', 'incident', 'prevalence', 'mortality', 'deaths', 'bound']
leakage_cols = [col for col in df.columns if any(word in col.lower() for word in forbidden_words)]
metadata_cols = [
    'Country or territory name', 'ISO 2-character country/territory code',
    'ISO 3-character country/territory code', 'ISO numeric country/territory code', 'Region'
]

# Create X by dropping target, leakage, and metadata
X = df.drop(columns=list(set(leakage_cols + metadata_cols + [target_col])), errors='ignore')
X = X.select_dtypes(include=[np.number]) # Keep only numeric predictors

# Grouped split
# Split by country name to ensure no country's data leaks into the test set
groups = df['Country or territory name']
gss = GroupShuffleSplit(test_size=0.3, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups))

X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

# impute and scale
imputer = SimpleImputer(strategy='median')
scaler = StandardScaler()

# Fit on training data only to avoid data leakage
X_train_processed = scaler.fit_transform(imputer.fit_transform(X_train))
X_test_processed = scaler.transform(imputer.transform(X_test))

# Fit Linear Regression model
lin_model = LinearRegression()
lin_model.fit(X_train_processed, y_train)

# evaluation and feature impact
y_pred = lin_model.predict(X_test_processed)

print(f"R-squared Score: {r2_score(y_test, y_pred):.4f}")
print(f"Mean Absolute Error: {mean_absolute_error(y_test, y_pred):.4f}")

# Check Top Predictors
importance = pd.DataFrame({
    'feature': X.columns,
    'weight': lin_model.coef_
}).sort_values(by='weight', ascending=False)

print("\nTop Predictors of Incidence Levels:")
print(importance.head(5))

# Visualization : Predicted vs Actual
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.5, color='teal')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2) # 45-degree line
plt.xlabel('Actual Incidence Rate')
plt.ylabel('Predicted Incidence Rate')
plt.title(f'Linear Regression: Predicted vs Actual (R²: {r2_score(y_test, y_pred):.2f})')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, classification_report
from sklearn.impute import SimpleImputer # Import SimpleImputer
import numpy as np # For np.number type check

# --- STEP 1: LOAD AND DEFINE TARGET ---
# 'high_tb' is already defined from previous cells.
y = df['high_tb']

# --- STEP 2: IDENTIFY LEAKAGE AND METADATA ---
# Explicitly list leakage columns that are direct indicators of the target or its components.
# The target is based on 'Estimated incidence (all forms) per 100 000 population'.
leakage_cols = [
    'Estimated incidence (all forms) per 100 000 population',
    'Estimated incidence (all forms) per 100 000 population, low bound',
    'Estimated incidence (all forms) per 100 000 population, high bound',
    'Estimated number of incident cases (all forms)',
    'Estimated number of incident cases (all forms), low bound',
    'Estimated number of incident cases (all forms), high bound',
    # All HIV-related incidence also strongly relates to overall incidence, considering them leakage
    'Estimated HIV in incident TB (percent)',
    'Estimated HIV in incident TB (percent), low bound',
    'Estimated HIV in incident TB (percent), high bound',
    'Estimated incidence of TB cases who are HIV-positive per 100 000 population',
    'Estimated incidence of TB cases who are HIV-positive per 100 000 population, low bound',
    'Estimated incidence of TB cases who are HIV-positive per 100 000 population, high bound',
    'Estimated incidence of TB cases who are HIV-positive',
    'Estimated incidence of TB cases who are HIV-positive, low bound',
    'Estimated incidence of TB cases who are HIV-positive, high bound',
    # Also removing prevalence and mortality per 100k, as they are very closely related to incidence
    'Estimated prevalence of TB (all forms) per 100 000 population',
    'Estimated prevalence of TB (all forms) per 100 000 population, low bound',
    'Estimated prevalence of TB (all forms) per 100 000 population, high bound',
    'Estimated mortality of TB cases (all forms, excluding HIV) per 100 000 population',
    'Estimated mortality of TB cases (all forms, excluding HIV), per 100 000 population, low bound',
    'Estimated mortality of TB cases (all forms, excluding HIV), per 100 000 population, high bound',
    'Estimated mortality of TB cases who are HIV-positive, per 100 000 population',
    'Estimated mortality of TB cases who are HIV-positive, per 100 000 population, low bound',
    'Estimated mortality of TB cases who are HIV-positive, per 100 000 population, high bound'
]

metadata_cols = [
    'Country or territory name',
    'ISO 2-character country/territory code',
    'ISO 3-character country/territory code',
    'ISO numeric country/territory code',
    'Region'
]

# Columns with very high missing values or purely descriptive that are not part of other drop lists
high_missing_or_descriptive_cols = [
    'Method to derive TBHIV estimates', # 100% missing
    'Method to derive incidence estimates',
    'Method to derive prevalence estimates',
    'Method to derive mortality estimates',
    'high_tb_burden' # This is just a duplicate of y, should not be in X
]

# Combine all columns to drop
cols_to_drop = list(set(leakage_cols + metadata_cols + high_missing_or_descriptive_cols + ['high_tb']))

# --- STEP 3: CREATE CLEAN X ---
# Ensure only existing columns are dropped to avoid errors
cols_to_drop = [col for col in cols_to_drop if col in df.columns]
X = df.drop(columns=cols_to_drop)

# Handle remaining categorical columns through one-hot encoding
X = pd.get_dummies(X, drop_first=True) # drop_first to avoid multicollinearity

# Impute missing numerical values after one-hot encoding
# Create imputer BEFORE splitting to avoid data leakage from test set
imputer = SimpleImputer(strategy='median')
X_numeric_cols = X.select_dtypes(include=np.number).columns
X[X_numeric_cols] = imputer.fit_transform(X[X_numeric_cols])

# Convert boolean columns (from get_dummies) to int, if any
for col in X.select_dtypes(include='bool').columns:
    X[col] = X[col].astype(int)

# --- STEP 4: RERUN THE SPLIT ---
# Using 'Country or territory name' for the groups, but 'Country or territory name' is dropped from X.
# We need to pass the original 'Country or territory name' from df to GroupShuffleSplit.
# This assumes that the indices of df and X are aligned, which they are if we only dropped columns.
groups = df.loc[X.index, 'Country or territory name'] # Use X.index to ensure alignment with X

gss = GroupShuffleSplit(test_size=0.3, random_state=42)

# Get indices and create new sets
# The split method expects X and y to be passed, along with groups.
# X here is the processed X with dummies and imputed values.
train_idx, test_idx = next(gss.split(X, y, groups))

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]
y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

# --- STEP 5: SCALE AND MODEL ---
scaler = StandardScaler()

# Check for remaining NaNs in train/test sets just in case (should be none after imputer)
if X_train.isnull().sum().sum() > 0:
    print("Warning: NaNs still present in X_train after imputation.")
if X_test.isnull().sum().sum() > 0:
    print("Warning: NaNs still present in X_test after imputation.")

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = LogisticRegression(max_iter=1000, solver='liblinear') # Use liblinear for smaller datasets or L1/L2 regularization
model.fit(X_train_scaled, y_train)

y_prob = model.predict_proba(X_test_scaled)[:, 1]
print(f"Realistic AUC: {roc_auc_score(y_test, y_prob):.4f}")

# Print detailed classification report
y_pred = model.predict(X_test_scaled)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Verify Feature importance
importance = pd.DataFrame({
    'feature': X_train.columns,
    'coefficient': model.coef_[0]
}).sort_values(by='coefficient', ascending=False)

print("\nTop 5 Features driving the model:")
print(importance.head(5))
```
